```r
# Load packages
library(readr)   # read_csv / write_csv
library(dplyr)   # data manipulation
library(readxl)  # read_excel / write_excel
```

```r
getwd()
list.files("raw_data")
facility_data_raw <- read_excel("raw_data/facility_data_raw.xlsx")
head(facility_data_raw)
```

```r
# explore columns
unique(facility_data_raw$Classification)
unique(facility_data_raw$`Operating Status`)

# check for missing data by row (there are none)
sum(!complete.cases(facility_data_raw))

# create new dataset for processed data
facility_data_processed <- facility_data_raw

head(facility_data_processed)
```

```r
# install and load library to change column names
names(facility_data_processed)
install.packages("janitor")
library(janitor)
facility_data_processed <- facility_data_processed %>%
  clean_names()
colnames(facility_data_processed)
```

```r
# filter for operating status of operating and classification of major emissions
facility_data_processed <- facility_data_processed %>%
  filter(operating_status == "Operating", classification == "Major Emissions")

# check duplicates by source_id and fiscal_year
facility_duplicate_check <- facility_data_processed %>%
  group_by(source_id, fiscal_year) %>%
  mutate(duplicate_check = if_else(n() > 1, "Duplicate", "Unique")) %>%
  ungroup()

sum(facility_duplicate_check$duplicate_check == "Duplicate", na.rm = TRUE)

# keep one row per facility and count active years
facility_data_processed <- facility_data_processed %>%
  group_by(state, source_id) %>%
  summarise(active_years = sum(active == 1, na.rm = TRUE), .groups = "drop") %>%
  filter(active_years == 10)

head(facility_data_processed)
```
